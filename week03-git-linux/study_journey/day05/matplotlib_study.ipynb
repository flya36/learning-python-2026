```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y1 = [2,7,12,9]
y2 = [5,10,15,20]

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] #字体为黑体

plt.figure(figsize =(10,5))
# 绘制折线图
plt.plot(x,y1,label = '折线1')
plt.plot(x,y2,color = 'yellow',label = '折线2')

# 设置折线图的详细信息
plt.title('折线图',fontsize = 20)   #标题
plt.xlabel('X轴',fontsize = 15)     #x轴标题
plt.ylabel('Y轴',fontsize = 15)     #y轴标题
plt.xticks(x)
y_ticks = [i for i in range(0,21)]
plt.yticks(y_ticks)
plt.grid(True,linestyle = '-',alpha = 0.6)
plt.legend(loc = 'upper right')

# 显示图表
plt.show()
```

```python
# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

# 创建子图
# figure: 画布，axes: 子图
figure, axes = plt.subplots(nrows=1, ncols=2,figsize=(10,5),dpi=100)

# 创建子图1：柱状图
countries = ['美国','中国','日本','印度','俄罗斯','法国','英国','德国','意大利','西班牙']
values = [2000, 1500, 1200, 900, 800, 700, 600, 500, 400, 300]
axes1:plt.Axes = axes[0]
axes1.bar(countries, values,color='red')
axes1.set_title('世界石油储备',fontsize=18)
axes1.set_xlabel('国家',fontsize=12)
axes1.set_ylabel('油量（亿吨）',fontsize=12)

# 创建子图2：饼状图（世界人口）
countries2 = ['美国','中国','日本','印度','俄罗斯','法国','英国','德国','意大利','西班牙']
values = [14.519, 13.02, 11.09, 9.09, 8.05, 7.07, 6.07, 5.05, 4.05, 3.05]

axes2:plt.Axes = axes[1]
axes2.pie(values, labels=countries2, autopct='%1.1f%%')
axes2.set_title('世界人口',fontsize=18)
axes2.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5, -0.13))
plt.show()
```
