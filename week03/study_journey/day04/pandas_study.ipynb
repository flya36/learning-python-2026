```python
import pandas as pd

# 创建DataFrame,学生成绩数据
df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    '语文': [80, 90, 70, 60],
    '数学': [80, 90, 70, 60],
    '英语': [80, 90, 70, 60]
})
df

# 统计计算
print(f'语文最高分：{df['语文'].max()},最低分：{df['语文'].min()},平均分：{df['语文'].mean()}')
print(f'数学最高分：{df['数学'].max()},最低分：{df['数学'].min()},平均分：{df['数学'].mean()}')
print(f'英语最高分：{df['英语'].max()},最低分：{df['英语'].min()},平均分：{df['英语'].mean()}')
```
